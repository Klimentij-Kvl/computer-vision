```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
import torch
import torch.nn as nn
```

```python
df = pd.read_csv("data/toxic_comments/train.csv")
df.head()
```

```python
train_df, val_df = train_test_split(
    df,
    test_size=0.1,
    random_state=42,
    shuffle=True,
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
```

```python
import re
from collections import Counter

def tokenize(text: str) -> list[str]:
    return re.findall(r"\w+|[^\w\s]", str(text).lower())

counter = Counter()

for text in train_df["comment_text"]:
    counter.update(tokenize(text))

PAD_TOKEN = "<PAD>"
UNK_TOKEN = "<UNK>"

PAD_IDX = 0
UNK_IDX = 1

MAX_VOCAB_SIZE = 50_000
MIN_FREQ = 2

vocab_tokens = [
    token
    for token, freq in counter.most_common(MAX_VOCAB_SIZE - 2)
    if freq >= MIN_FREQ
]

itos = [PAD_TOKEN, UNK_TOKEN] + vocab_tokens
stoi = {token: idx for idx, token in enumerate(itos)}

print("Vocabulary size:", len(stoi))
```
